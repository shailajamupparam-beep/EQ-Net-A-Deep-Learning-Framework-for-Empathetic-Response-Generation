```python
%%writefile eqnet_plus/modules/data_loader.py
import sys
import os
from pathlib import Path
from huggingface_hub import hf_hub_download
from datasets import load_dataset, Features, Value
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

# Add the parent directory of eqnet_plus to sys.path
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

from eqnet_plus.config import cfg

def load_empathetic_dialogues():
    """
    Load EmpatheticDialogues from HuggingFace Hub by downloading and then loading local JSON files.
    """
    print("Downloading EmpatheticDialogues JSON files from HuggingFace Hub...")

    repo_id = "facebook/empathetic_dialogues"

    # Corrected dictionary with clear file path mappings
    data_files = {
        'train': 'data/data_json/train.json',
        'validation': 'data/data_json/validation.json',
        'test': 'data/data_json/test.json'
    }

    local_data_paths = {}
    try:
        # Check config object or fallback safely
        download_dir = cfg.DATA_RAW if (cfg and hasattr(cfg, 'DATA_RAW')) else "eqnet_plus/data/raw"
        os.makedirs(download_dir, exist_ok=True)

        for split, file_path in data_files.items():
            print(f"Downloading {split} split...")
            local_file = hf_hub_download(repo_id=repo_id, filename=file_path, repo_type="dataset")
            local_data_paths[split] = local_file

        # Load dataset from the downloaded local JSON files
        dataset = load_dataset('json', data_files=local_data_paths)
        print("Dataset loaded successfully! ✓")
        return dataset

    except Exception as e:
        print(f"An error occurred during download/loading: {e}")
        print("Falling back to standard direct API loading...")
        return load_dataset(repo_id, trust_remote_code=True)

if __name__ == "__main__":
    dataset = load_empathetic_dialogues()
    print(dataset)
```

```python
%%writefile eqnet_plus/modules/data_loader.py
import sys
import os
from pathlib import Path
import requests # Added requests for direct download
from datasets import load_dataset, Features, Value # Keep Features and Value
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

# Add parent directory to sys.path so we can import local config
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

try:
    from eqnet_plus.config import cfg
    print("Config imported successfully ✓")
except ImportError:
    print("Could not import eqnet_plus.config, using default paths.")
    class MockConfig: # Define a mock config for safety
        DATA_RAW = Path("eqnet_plus/data/raw")
    cfg = MockConfig()
    os.makedirs(cfg.DATA_RAW, exist_ok=True) # Ensure path exists if using mock config

def load_empathetic_dialogues():
    """
    Load EmpatheticDialogues from HuggingFace Hub by directly downloading JSON files
    and then loading them locally using the datasets library.
    """
    print("Downloading EmpatheticDialogues JSON files from HuggingFace Hub...")

    # Corrected base_remote_url to include 'data_json/' with '?download=true' for direct file access
    base_remote_url = "https://huggingface.co/datasets/facebook/empathetic_dialogues/resolve/main/data/data_json/"
    data_file_names = {
        'train': "train.json",
        'validation': "validation.json",
        'test': "test.json",
    }

    local_data_paths = {}
    download_dir = cfg.DATA_RAW
    os.makedirs(download_dir, exist_ok=True)

    try:
        for split, file_name in data_file_names.items():
            # Append ?download=true as required by Hugging Face for direct download
            remote_url = f"{base_remote_url}{file_name}?download=true"
            local_path = download_dir / file_name
            print(f"Downloading {remote_url} to {local_path}...")
            response = requests.get(remote_url, stream=True)
            response.raise_for_status() # Raise an exception for HTTP errors
            with open(local_path, 'wb') as f:
                for chunk in response.iter_content(chunk_size=8192):
                    f.write(chunk)
            local_data_paths[split] = str(local_path)
            print(f"Downloaded {file_name} successfully.")

    except requests.exceptions.RequestException as e:
        raise RuntimeError(f"Failed to download EmpatheticDialogues JSON files using requests: {e}")

    print("Loading local JSON files using datasets library...")

    # Define the features explicitly for empathetic_dialogues to aid schema inference
    features = Features({
        'conv_id': Value('string'),
        'speaker_idx': Value('int32'),
        'utterance': Value('string'),
        'context': Value('string'),
        'prompt': Value('string'),
    })

    try:
        dataset = load_dataset(
            'json',
            data_files=local_data_paths,
            features=features # Explicitly pass features to help schema inference
        )
        print("Dataset loaded successfully! ✓")
        return dataset

    except Exception as e:
        raise RuntimeError(f"Failed to load dataset from local JSON files using datasets library: {e}")

if __name__ == "__main__":
    dataset = load_empathetic_dialogues()
    print("\nDataset Structure:")
    print(dataset)
```

```python
# modules/data_loader.py

import os
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# Create folders if they don't exist
os.makedirs("logs", exist_ok=True)
os.makedirs("data/raw", exist_ok=True)


def load_empathetic_dialogues():
    """
    Load EmpatheticDialogues dataset from Hugging Face.
    """

    print("Downloading EmpatheticDialogues dataset...")

    try:
        dataset = load_dataset("empathetic_dialogues")

        train_df = dataset["train"].to_pandas()
        val_df = dataset["validation"].to_pandas()
        test_df = dataset["test"].to_pandas()

        print(f"Train: {len(train_df)}")
        print(f"Validation: {len(val_df)}")
        print(f"Test: {len(test_df)}")

        return train_df, val_df, test_df

    except Exception as e:
        print("Error while downloading dataset:")
        print(e)
        return None, None, None


def explore_dataset(df, split_name="Train"):
    """
    Perform basic EDA on dataset.
    """

    if df is None:
        print("Dataset not loaded.")
        return

    print("\n===============================")
    print(f"{split_name} Dataset")
    print("===============================")

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 3 Rows:")
    print(df.head(3))

    # Emotion distribution
    emotion_counts = Counter(df["context"])

    print("\nUnique emotions:", len(emotion_counts))
    print("\nTop 10 emotions:")

    for emotion, count in emotion_counts.most_common(10):
        print(f"{emotion}: {count}")

    # Utterance length
    df["utt_len"] = df["utterance"].astype(str).apply(
        lambda x: len(x.split())
    )

    print("\nAverage utterance length:",
          round(df["utt_len"].mean(), 2))

    print("Maximum length:", df["utt_len"].max())
    print("Minimum length:", df["utt_len"].min())

    # Plotting
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))

    # Emotion distribution
    pd.Series(emotion_counts).sort_values(
        ascending=False
    ).plot(
        kind="bar",
        ax=axes[0],
        color="mediumpurple"
    )

    axes[0].set_title("Emotion Distribution")
    axes[0].set_xlabel("Emotion")
    axes[0].set_ylabel("Count")
    axes[0].tick_params(axis="x", rotation=45)

    # Utterance length
    axes[1].hist(
        df["utt_len"],
        bins=40,
        color="skyblue",
        edgecolor="black"
    )

    axes[1].set_title("Utterance Length Distribution")
    axes[1].set_xlabel("Words")
    axes[1].set_ylabel("Frequency")

    plt.tight_layout()

    save_path = "logs/eda_plots.png"
    plt.savefig(save_path, dpi=150)
    plt.close()

    print(f"\nEDA plot saved to {save_path}")


if __name__ == "__main__":

    train_df, val_df, test_df = load_empathetic_dialogues()

    if train_df is not None:

        explore_dataset(train_df, "Train")

        train_df.to_csv("data/raw/train.csv", index=False)
        val_df.to_csv("data/raw/val.csv", index=False)
        test_df.to_csv("data/raw/test.csv", index=False)

        print("\nDatasets saved successfully.")
```

```python
%%writefile eqnet_plus/modules/preprocessor.py
# modules/preprocessor.py
import re, json
import pandas as pd
import nltk, spacy
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

import sys
# Add the parent directory of eqnet_plus to sys.path
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

from eqnet_plus.config import cfg

nlp = spacy.load("en_core_web_sm")
STOP_WORDS = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

# High-arousal negative emotions → HIGH intensity
HIGH_INTENSITY_EMOTIONS = {
    "terrified", "furious", "devastated", "afraid", "angry"
}
# Medium intensity
MED_INTENSITY_EMOTIONS = {
    "anxious", "apprehensive", "sad", "disappointed",
    "lonely", "guilty", "embarrassed", "jealous"
}

def clean_text(text: str) -> str:
    """Basic text cleaning."""
    text = text.lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^a-z0-9\s'.,!?]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

def extract_emotional_keywords(text: str) -> list:
    """Extract nouns, adjectives, verbs as emotional keywords using SpaCy."""
    doc = nlp(text)
    keywords = [
        token.lemma_ for token in doc
        if token.pos_ in ("NOUN", "ADJ", "VERB")
        and token.text not in STOP_WORDS
        and len(token.text) > 2
    ]
    return keywords

def assign_intensity_label(emotion: str) -> str:
    """Pseudo-label intensity from emotion category."""
    emotion = emotion.lower().strip()
    if emotion in HIGH_INTENSITY_EMOTIONS:
        return "high"
    elif emotion in MED_INTENSITY_EMOTIONS:
        return "medium"
    else:
        return "low"

def assign_empathy_strategy(emotion: str, intensity: str) -> str:
    """Rule-based empathy strategy assignment."""
    if intensity == "high":
        return "validation"
    elif emotion in {"sad", "lonely", "disappointed"}:
        return "encouragement"
    elif emotion in {"anxious", "apprehensive"}:
        return "reassurance"
    elif emotion in {"excited", "hopeful", "grateful"}:
        return "acknowledgement"
    else:
        return "suggestion"

def preprocess_dataframe(df: pd.DataFrame) -> pd.DataFrame:
    """Full preprocessing pipeline on a split dataframe."""
    df = df.copy()
    df["utterance_clean"] = df["utterance"].apply(clean_text)
    df["keywords"] = df["utterance_clean"].apply(extract_emotional_keywords)
    df["intensity"] = df["context"].apply(assign_intensity_label)
    df["strategy"] = df.apply(lambda r: assign_empathy_strategy(
        r["context"], r["intensity"]), axis=1)
    return df

if __name__ == "__main__":
    train_df = pd.read_csv("data/raw/train.csv")
    val_df   = pd.read_csv("data/raw/val.csv")
    test_df  = pd.read_csv("data/raw/test.csv")

    for name, df in [("train", train_df), ("val", val_df), ("test", test_df)]:
        proc = preprocess_dataframe(df)
        proc.to_csv(f"data/processed/{name}_proc.csv", index=False)
        print(f"{name} processed: {len(proc)} rows ✓")
```

```python
# modules/dataset.py
import torch
from torch.utils.data import Dataset, DataLoader
import pandas as pd
import json
from transformers import BertTokenizer

import sys
# Add the parent directory of eqnet_plus to sys.path
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

from eqnet_plus.config import cfg

# Ensure the label_maps.json path is correctly constructed using cfg.DATA_PROC
with open(cfg.DATA_PROC / "label_maps.json") as f:
    MAPS = json.load(f)

tokenizer = BertTokenizer.from_pretrained(cfg.ENCODER_MODEL)

class EQNetDataset(Dataset):
    def __init__(self, csv_path: str):
        # Keep default NA values as False to prevent empty strings from being read as NaN
        # and subsequently dropped, leading to num_samples=0 error.
        self.df = pd.read_csv(csv_path, keep_default_na=False).dropna(subset=["utterance_clean"])
        self.df = self.df.reset_index(drop=True)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        utterance = str(row["utterance_clean"])
        response  = str(row.get("utterance", utterance))  # target response

        # Tokenize input
        enc = tokenizer(
            utterance,
            max_length=cfg.MAX_LEN,
            padding="max_length",
            truncation=True,
            return_tensors="pt"
        )

        # Tokenize target response (for decoder)
        tgt = tokenizer(
            response,
            max_length=cfg.MAX_LEN,
            padding="max_length",
            truncation=True,
            return_tensors="pt"
        )

        # Labels
        emotion_id  = MAPS["emotion2id"].get(str(row["emotion"]).lower(), 0) # Corrected to use 'emotion' column
        intensity_id= MAPS["intensity2id"].get(str(row["intensity"]), 0)
        strategy_id = MAPS["strategy2id"].get(str(row["strategy"]), 0)

        return {
            "input_ids":      enc["input_ids"].squeeze(),
            "attention_mask": enc["attention_mask"].squeeze(),
            "token_type_ids": enc["token_type_ids"].squeeze(),
            "decoder_ids":    tgt["input_ids"].squeeze(),
            "emotion_label":  torch.tensor(emotion_id,   dtype=torch.long),
            "intensity_label":torch.tensor(intensity_id, dtype=torch.long),
            "strategy_label": torch.tensor(strategy_id,  dtype=torch.long),
        }

def get_dataloaders():
    train_ds = EQNetDataset(str(cfg.DATA_PROC / "train_proc.csv"))
    val_ds   = EQNetDataset(str(cfg.DATA_PROC / "val_proc.csv"))
    test_ds  = EQNetDataset(str(cfg.DATA_PROC / "test_proc.csv"))

    train_loader = DataLoader(train_ds, batch_size=cfg.BATCH_SIZE,
                               shuffle=True, num_workers=4, pin_memory=True)
    val_loader   = DataLoader(val_ds,   batch_size=cfg.BATCH_SIZE, shuffle=False)
    test_loader  = DataLoader(test_ds,  batch_size=cfg.BATCH_SIZE, shuffle=False)

    print(f"Train batches: {len(train_loader)} | Val: {len(val_loader)} | Test: {len(test_loader)}")
    return train_loader, val_loader, test_loader

if __name__ == "__main__":
    tl, vl, tel = get_dataloaders()
    batch = next(iter(tl))
    for k, v in batch.items():
        print(f"{k}: {v.shape}")
```

```python
# This cell is designed to rewrite and then execute preprocessor.py
# to ensure the processed CSV files are created.

%%writefile eqnet_plus/modules/preprocessor.py
# modules/preprocessor.py
import re, json
import pandas as pd
import nltk, spacy
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import sys # Added import for sys
import os # Added import for os

# Add the parent directory of eqnet_plus to sys.path if not already present
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

from eqnet_plus.config import cfg # Corrected import statement

# Ensure NLTK data is downloaded
# ==============================
nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)

try:
    nlp = spacy.load("en_core_web_sm")
except:
    os.system("python -m spacy download en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")

STOP_WORDS = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

# High-arousal negative emotions → HIGH intensity
HIGH_INTENSITY_EMOTIONS = {
    "terrified", "furious", "devastated", "afraid", "angry"
}
# Medium intensity
MED_INTENSITY_EMOTIONS = {
    "anxious", "apprehensive", "sad", "disappointed",
    "lonely", "guilty", "embarrassed", "jealous"
}

def clean_text(text: str) -> str:
    """Basic text cleaning."""
    text = text.lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^a-z0-9\s'.,!?]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

def extract_emotional_keywords(text: str) -> list:
    """Extract nouns, adjectives, verbs as emotional keywords using SpaCy."""
    doc = nlp(text)
    keywords = [
        token.lemma_ for token in doc
        if token.pos_ in ("NOUN", "ADJ", "VERB")
        and token.text not in STOP_WORDS
        and len(token.text) > 2
    ]
    return keywords

def assign_intensity_label(emotion: str) -> str:
    """Pseudo-label intensity from emotion category."""
    emotion = emotion.lower().strip()
    if emotion in HIGH_INTENSITY_EMOTIONS:
        return "high"
    elif emotion in MED_INTENSITY_EMOTIONS:
        return "medium"
    else:
        return "low"

def assign_empathy_strategy(emotion: str, intensity: str) -> str:
    """Rule-based empathy strategy assignment."""
    if intensity == "high":
        return "validation"
    elif emotion in {"sad", "lonely", "disappointed"}:
        return "encouragement"
    elif emotion in {"anxious", "apprehensive"}:
        return "reassurance"
    elif emotion in {"excited", "hopeful", "grateful"}:
        return "acknowledgement"
    else:
        return "suggestion"

def preprocess_dataframe(df: pd.DataFrame) -> pd.DataFrame:
    """Full preprocessing pipeline on a split dataframe."""
    df = df.copy()

    # The 'conversations' column is already a list of dicts from data_loader.py
    # No need to call json.loads()

    # Extract the last utterance content from the 'conversations' list
    def get_last_utterance_content(conv_list):
        if conv_list and isinstance(conv_list, list) and isinstance(conv_list[-1], dict) and 'content' in conv_list[-1]:
            return conv_list[-1]['content']
        return ""

    df["utterance"] = df["conversations"].apply(get_last_utterance_content)

    df["utterance_clean"] = df["utterance"].apply(clean_text)
    df["keywords"] = df["utterance_clean"].apply(extract_emotional_keywords)
    # Corrected: Use 'emotion' column instead of 'context'
    df["intensity"] = df["emotion"].apply(assign_intensity_label)
    df["strategy"] = df.apply(lambda r: assign_empathy_strategy(
        r["emotion"], r["intensity"]), axis=1) # Corrected: Use 'emotion' column
    return df

def run_preprocessor_pipeline():
    print("Running preprocessor pipeline...")
    # Ensure the required directories exist
    os.makedirs(cfg.DATA_RAW, exist_ok=True)
    os.makedirs(cfg.DATA_PROC, exist_ok=True)
    os.makedirs(os.path.join(cfg.BASE_DIR, "modules"), exist_ok=True) # Ensure modules dir exists in relation to BASE_DIR

    # Print the resolved raw data path
    print(f"Preprocessor expects raw data at: {cfg.DATA_RAW}")

    # Load the in-memory DataFrames from raw CSVs (which should have been created by data_loader.py)
    try:
        print(f"DEBUG: Attempting to load raw train data from: {cfg.DATA_RAW / 'train.csv'}")
        train_df_loaded = pd.read_csv(cfg.DATA_RAW / "train.csv")
        print(f"DEBUG: Attempting to load raw val data from: {cfg.DATA_RAW / 'val.csv'}")
        val_df_loaded   = pd.read_csv(cfg.DATA_RAW / "val.csv")
        print(f"DEBUG: Attempting to load raw test data from: {cfg.DATA_RAW / 'test.csv'}")
        test_df_loaded  = pd.read_csv(cfg.DATA_RAW / "test.csv")
    except FileNotFoundError:
        print(f"Error: Raw data CSVs not found. Please ensure data_loader.py is executed to generate them in {cfg.DATA_RAW}.")
        exit() # Exit if raw data isn't available


    for name, df in [("train", train_df_loaded), ("val", val_df_loaded), ("test", test_df_loaded)]:
        proc = preprocess_dataframe(df)
        output_path = cfg.DATA_PROC / f"{name}_proc.csv"
        print(f"DEBUG: Saving processed {name} data to: {output_path}")
        proc.to_csv(str(output_path), index=False) # Converted Path to string
        print(f"{name} processed: {len(proc)} rows \u2713")
    print("Preprocessor pipeline complete.")

if __name__ == "__main__":
    run_preprocessor_pipeline()
```

```python
class EQNetDataset(Dataset):

    def __init__(self, csv_path):

        self.df = pd.read_csv(csv_path)

        print("Columns:", self.df.columns.tolist())

        self.df = self.df.fillna("")

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        utterance = str(row["utterance_clean"])
        response = str(row["utterance"])

        enc = tokenizer(
            utterance,
            padding="max_length",
            truncation=True,
            max_length=cfg.MAX_LEN,
            return_tensors="pt"
        )

        dec = tokenizer(
            response,
            padding="max_length",
            truncation=True,
            max_length=cfg.MAX_LEN,
            return_tensors="pt"
        )

        emotion = MAPS["emotion2id"].get(
            str(row["emotion"]).lower(), 0
        )

        intensity = MAPS["intensity2id"].get(
            str(row["intensity"]).lower(), 0
        )

        strategy = MAPS["strategy2id"].get(
            str(row["strategy"]).lower(), 0
        )

        return {

            "input_ids": enc["input_ids"].squeeze(),

            "attention_mask": enc["attention_mask"].squeeze(),

            "decoder_ids": dec["input_ids"].squeeze(),

            "emotion_label": torch.tensor(emotion),

            "intensity_label": torch.tensor(intensity),

            "strategy_label": torch.tensor(strategy)
        }
```

```python
import os
import sys
import torch
import torch.nn as nn
import torch.optim as optim
from transformers import BertModel
from pathlib import Path

# Add the parent directory of eqnet_plus to sys.path if not already present
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

# Ensure the eqnet_plus package structure with __init__.py files
os.makedirs('/content/eqnet_plus', exist_ok=True)
os.makedirs('/content/eqnet_plus/modules', exist_ok=True)
open('/content/eqnet_plus/__init__.py', 'a').close()
open('/content/eqnet_plus/modules/__init__.py', 'a').close()

# --- DIRECTLY DEFINE CONFIG CLASS TO BYPASS PERSISTENT IMPORT ERROR ---
class Config:
    # ── Paths
    BASE_DIR = Path("/content/eqnet_plus") # Explicitly set base path
    DATA_RAW = BASE_DIR / "data/raw"
    DATA_PROC = BASE_DIR / "data/processed"
    MODEL_DIR = BASE_DIR / "models"

    # ── Model names
    ENCODER_MODEL = "bert-base-uncased"
    DECODER_MODEL = "microsoft/DialoGPT-small"

    # ── Emotion labels (EmpatheticDialogues 32 emotions)
    EMOTION_LABELS = [
        "admiring", "afraid", "angry", "annoyed", "anticipating",
        "anxious", "apprehensive", "ashamed", "caring", "confident",
        "content", "devastated", "disappointed", "disgusted", "embarrassed",
        "excited", "faithful", "furious", "grateful", "guilty",
        "hopeful", "impressed", "jealous", "joyful", "lonely",
        "nostalgic", "prepared", "proud", "sad", "sentimental",
        "surprised", "terrified"
    ]
    NUM_EMOTIONS = 32

    # ── Intensity levels
    INTENSITY_LABELS = ["low", "medium", "high"]
    NUM_INTENSITY = 3

    # ── Empathy strategies
    EMPATHY_STRATEGIES = [
        "acknowledgement", "encouragement", "suggestion",
        "validation", "reassurance"
    ]
    NUM_STRATEGIES = 5

    # ── Training
    MAX_LEN = 128
    BATCH_SIZE = 16
    EPOCHS = 5
    LR = 2e-5
    WARMUP_STEPS = 500
    SEED = 42

    # ── Multi-task loss weights
    LOSS_W_EMOTION = 0.4
    LOSS_W_INTENSITY = 0.3
    LOSS_W_STRATEGY = 0.3

    # ── Device
    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

cfg = Config()
# --- END DIRECT CONFIG DEFINITION ---


class EmotionDetector(nn.Module):
    """
    BERT + Classification Head
    """

    def __init__(self,
                 num_labels=cfg.NUM_EMOTIONS,
                 dropout=0.3):

        super(EmotionDetector, self).__init__()

        self.bert = BertModel.from_pretrained(cfg.ENCODER_MODEL)

        hidden_size = self.bert.config.hidden_size

        self.dropout = nn.Dropout(dropout)

        self.classifier = nn.Sequential(

            nn.Linear(hidden_size, 256),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(256, num_labels)

        )

    def forward(
        self,
        input_ids,
        attention_mask,
        token_type_ids=None
    ):

        outputs = self.bert(

            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids

        )

        cls = outputs.last_hidden_state[:, 0]

        cls = self.dropout(cls)

        logits = self.classifier(cls)

        return logits, cls


########################################################################
# TRAINING FUNCTION
########################################################################

def train_emotion_detector(
        train_loader,
        val_loader=None,
        epochs=cfg.EPOCHS):

    device = cfg.DEVICE

    model = EmotionDetector().to(device)

    optimizer = optim.AdamW(
        model.parameters(),
        lr=cfg.LR
    )

    criterion = nn.CrossEntropyLoss()

    best_acc = 0

    for epoch in range(epochs):

        model.train()

        total_loss = 0
        correct = 0
        total = 0

        for batch in train_loader:

            input_ids = batch[
                "input_ids"].to(device)

            attention_mask = batch[
                "attention_mask"].to(device)

            labels = batch["emotion_label"].to(device)

            token_type_ids = batch.get("token_type_ids", None)

            if token_type_ids is not None:
                token_type_ids = token_type_ids.to(device)

            optimizer.zero_grad()

            logits, _ = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                token_type_ids=token_type_ids
            )

            loss = criterion(logits, labels)

            loss.backward()

            optimizer.step()

            total_loss += loss.item()

            predictions = torch.argmax(logits, dim=1)

            correct += (predictions == labels).sum().item()

            total += labels.size(0)

        train_acc = correct / total

        print("=" * 50)

        print(f"Epoch : {epoch+1}/{epochs}")

        print(f"Loss  : {total_loss/len(train_loader):.4f}")

        print(f"Train Accuracy : {train_acc:.4f}")

        ###############################################################
        # Validation
        ###############################################################

        if val_loader is not None:

            model.eval()

            val_correct = 0

            val_total = 0

            with torch.no_grad():

                for batch in val_loader:

                    input_ids = batch["input_ids"].to(device)

                    attention_mask = batch["attention_mask"].to(device)

                    labels = batch["emotion_label"].to(device)

                    token_type_ids = batch.get("token_type_ids", None)

                    if token_type_ids is not None:
                        token_type_ids = token_type_ids.to(device)

                    logits, _ = model(

                        input_ids=input_ids,
                        attention_mask=attention_mask,
                        token_type_ids=token_type_ids

                    )

                    pred = torch.argmax(logits, dim=1)

                    val_correct += (pred == labels).sum().item()

                    val_total += labels.size(0)

            val_acc = val_correct / val_total

            print(f"Validation Accuracy : {val_acc:.4f}")

            if val_acc > best_acc:
                best_acc = val_acc

                os.makedirs("models/emotion_detector", exist_ok=True)

                torch.save(
                    model.state_dict(),
                    "models/emotion_detector/model.pt"
                )

                print("Best Model Saved")

    os.makedirs("models/emotion_detector", exist_ok=True)

    torch.save(

        model.state_dict(),

        "models/emotion_detector/model.pt"

    )

    print("\nEmotion Detector Saved Successfully")

    return model
```

```python
# modules/cause_model.py
import spacy
import re

nlp = spacy.load("en_core_web_sm")

# Trigger words that often precede a causal phrase
CAUSE_TRIGGERS = [
    "because", "since", "as", "due to", "owing to",
    "when", "after", "caused by", "i lost", "i failed",
    "my", "i was", "she", "he"
]

def extract_cause_rule_based(text: str) -> str:
    """Pattern-based cause extraction."""
    text_lower = text.lower()
    for trigger in CAUSE_TRIGGERS:
        idx = text_lower.find(trigger)
        if idx != -1:
            fragment = text[idx:idx+60].strip()
            # Cut at sentence boundary
            fragment = re.split(r"[.!?]", fragment)[0]
            if len(fragment) > 5:
                return fragment
    return "unspecified event"

def extract_cause_spacy(text: str) -> str:
    """SpaCy dependency parsing approach for cause extraction."""
    doc = nlp(text)

    # Look for advcl (adverbial clause) or ccomp (clausal complement)
    for token in doc:
        if token.dep_ in ("advcl", "ccomp", "relcl"):
            subtree = " ".join([t.text for t in token.subtree])
            if len(subtree.split()) > 2:
                return subtree[:80]

    # Fallback: return the subject + verb phrase
    subject, verb = "", ""
    for token in doc:
        if "subj" in token.dep_:
            subject = token.text
        if token.pos_ == "VERB" and not verb:
            verb = token.text
    if subject and verb:
        return f"{subject} {verb}"
    return extract_cause_rule_based(text)

def identify_cause(text: str) -> dict:
    """Main API: returns cause phrase and named entities."""
    doc = nlp(text)
    entities = [(ent.text, ent.label_) for ent in doc.ents]
    cause_phrase = extract_cause_spacy(text)
    return {"cause": cause_phrase, "entities": entities}

if __name__ == "__main__":
    test_texts = [
        "I am so sad because I lost my job today.",
        "I feel anxious when I have to speak in public.",
        "She broke up with me and I'm devastated."
    ]
    for t in test_texts:
        result = identify_cause(t)
        print(f"Text: {t}")
        print(f"Cause: {result['cause']} | Entities: {result['entities']}\n")
```

```python
import torch
import torch.nn as nn
from transformers import BertModel
from eqnet_plus.modules.intensity_model import IntensityHead
from eqnet_plus.modules.empathy_model import EmpathyHead
from eqnet_plus.config import cfg

class EQNetPlus(nn.Module):
    """
    EQ-Net+: Shared BERT encoder with 3 task heads.
    Emotion Detection | Intensity Estimation | Empathy Strategy
    """
    def __init__(self):
        super().__init__()
        # Shared backbone
        self.bert = BertModel.from_pretrained(cfg.ENCODER_MODEL)
        self.dropout = nn.Dropout(0.3)
        hidden = self.bert.config.hidden_size  # 768

        # Task heads
        self.emotion_head = nn.Sequential(
            nn.Linear(hidden, 256), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(256, cfg.NUM_EMOTIONS)
        )
        self.intensity_head = IntensityHead(hidden)
        self.empathy_head   = EmpathyHead(hidden)

    def forward(self, input_ids, attention_mask, token_type_ids=None,
               intensity_ids=None):
        # Shared encoding
        bert_out = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        cls_emb = self.dropout(bert_out.last_hidden_state[:, 0, :])

        # Individual heads
        emotion_logits   = self.emotion_head(cls_emb)
        intensity_logits = self.intensity_head(cls_emb)

        # Empathy head uses predicted intensity if not provided
        if intensity_ids is None:
            intensity_ids = intensity_logits.argmax(dim=-1)
        empathy_logits = self.empathy_head(cls_emb, intensity_ids)

        return {
            "emotion_logits":   emotion_logits,
            "intensity_logits": intensity_logits,
            "empathy_logits":   empathy_logits,
            "cls_emb":          cls_emb
        }


def compute_multitask_loss(outputs, labels):
    """Weighted multi-task loss."""
    ce = nn.CrossEntropyLoss()
    l_emotion   = ce(outputs["emotion_logits"],   labels["emotion"])
    l_intensity = ce(outputs["intensity_logits"], labels["intensity"])
    l_empathy   = ce(outputs["empathy_logits"],   labels["strategy"])

    total = (cfg.LOSS_W_EMOTION   * l_emotion +
             cfg.LOSS_W_INTENSITY * l_intensity +
             cfg.LOSS_W_STRATEGY  * l_empathy)
    return total, l_emotion, l_intensity, l_empathy

if __name__ == "__main__":
    # Ensure sys.path is correctly configured for local imports
    import sys
    if '/content' not in sys.path:
        sys.path.insert(0, '/content')

    # Force reload of eqnet_plus.config to pick up any changes
    import importlib
    if 'eqnet_plus.config' in sys.modules:
        importlib.reload(sys.modules['eqnet_plus.config'])
    from eqnet_plus.config import cfg # Re-import cfg if running directly

    model = EQNetPlus().to(cfg.DEVICE)
    print(f"EQNet+ params: {sum(p.numel() for p in model.parameters()):,}")
    # Quick forward pass test
    dummy_ids  = torch.randint(0, 30522, (2, cfg.MAX_LEN)).to(cfg.DEVICE)
    dummy_mask = torch.ones(2, cfg.MAX_LEN, dtype=torch.long).to(cfg.DEVICE)
    out = model(dummy_ids, dummy_mask)
    for k, v in out.items():
        print(f"{k}: {v.shape}")
```

```python
%%writefile /content/eqnet_plus/modules/response_model.py
import torch
import torch.nn as nn
from transformers import GPT2LMHeadModel, GPT2Tokenizer

import sys
# Add the parent directory of eqnet_plus to sys.path
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

from eqnet_plus.config import cfg # Corrected import

gpt_tokenizer = GPT2Tokenizer.from_pretrained(cfg.DECODER_MODEL)
gpt_tokenizer.pad_token = gpt_tokenizer.eos_token

class EmpathyResponseGenerator(nn.Module):
    """
    Conditioned DialoGPT for empathetic response generation.
    Prefix = emotion_embed + strategy_embed → GPT-2 LM head.
    """
    def __init__(self):
        super().__init__()
        self.gpt = GPT2LMHeadModel.from_pretrained(cfg.DECODER_MODEL)
        emb_dim = self.gpt.config.n_embd  # 768 for DialoGPT-small

        # Prefix embeddings for emotion and strategy
        self.emotion_prefix   = nn.Embedding(cfg.NUM_EMOTIONS,   emb_dim)
        self.strategy_prefix  = nn.Embedding(cfg.NUM_STRATEGIES, emb_dim)
        self.prefix_proj      = nn.Linear(emb_dim * 2, emb_dim)

    def forward(self, input_ids, attention_mask, labels,
               emotion_ids, strategy_ids):
        B = input_ids.size(0)
        # Build prefix embedding: [emotion | strategy] → project to GPT emb dim
        em  = self.emotion_prefix(emotion_ids)    # [B, D]
        st  = self.strategy_prefix(strategy_ids)  # [B, D]
        prefix = self.prefix_proj(
            torch.cat([em, st], dim=-1)
        ).unsqueeze(1)  # [B, 1, D]

        # GPT token embeddings
        tok_emb = self.gpt.transformer.wte(input_ids)      # [B, L, D]
        pos_emb = self.gpt.transformer.wpe(
            torch.arange(input_ids.size(1)+1, device=input_ids.device)
        ).unsqueeze(0)
        # Concat prefix token before input
        full_emb = torch.cat([prefix, tok_emb], dim=1) + pos_emb  # [B, L+1, D]
        prefix_mask = torch.ones(B, 1, device=attention_mask.device,
                                   dtype=attention_mask.dtype)
        full_mask = torch.cat([prefix_mask, attention_mask], dim=1)

        # Pad labels to match length
        pad = torch.full((B, 1), -100, dtype=labels.dtype, device=labels.device)
        full_labels = torch.cat([pad, labels], dim=1)

        out = self.gpt(inputs_embeds=full_emb, attention_mask=full_mask,
                       labels=full_labels)
        return out.loss

    @torch.no_grad()
    def generate_response(self, utterance: str, emotion_id: int,
                          strategy_id: int, max_new_tokens=80) -> str:
        self.eval()
        enc = gpt_tokenizer(utterance, return_tensors="pt").to(cfg.DEVICE)
        em  = self.emotion_prefix(torch.tensor([emotion_id]).to(cfg.DEVICE))
        st  = self.strategy_prefix(torch.tensor([strategy_id]).to(cfg.DEVICE))
        prefix = self.prefix_proj(
            torch.cat([em, st], dim=-1)
        ).unsqueeze(0)

        tok_emb = self.gpt.transformer.wte(enc["input_ids"])
        full_emb = torch.cat([prefix, tok_emb], dim=1)
        pos_emb = self.gpt.transformer.wpe(
            torch.arange(full_emb.size(1), device=cfg.DEVICE)
        ).unsqueeze(0)
        full_emb = full_emb + pos_emb

        out = self.gpt.generate(
            inputs_embeds=full_emb,
            max_new_tokens=max_new_tokens,
            do_sample=True, top_p=0.92, temperature=0.8,
            pad_token_id=gpt_tokenizer.eos_token_id
        )
        return gpt_tokenizer.decode(out[0], skip_special_tokens=True)
```

```python
# evaluation/metrics.py
import torch
import math
from evaluate import load
from sklearn.metrics import accuracy_score, classification_report
from transformers import GPT2LMHeadModel, GPT2Tokenizer
from config import cfg

bleu = load("sacrebleu")

# Empathy keywords per strategy
EMPATHY_KEYWORDS = {
    "validation":     ["understand", "makes sense", "valid", "feel"],
    "encouragement":  ["believe in you", "keep going", "strong", "you can"],
    "reassurance":    ["alright", "okay", "be fine", "safe"],
    "acknowledgement":["that", "wonderful", "great", "happy for you"],
    "suggestion":     ["try", "maybe", "could", "consider"],
}

def compute_bleu(predictions: list, references: list) -> float:
    """Compute corpus BLEU score."""
    refs_wrapped = [[r] for r in references]
    result = bleu.compute(predictions=predictions, references=refs_wrapped)
    return result["score"]

def compute_perplexity(model, tokenizer, texts: list) -> float:
    """Compute average perplexity on a list of texts."""
    model.eval()
    total_log_prob, total_tokens = 0, 0
    with torch.no_grad():
        for text in texts:
            enc = tokenizer(text, return_tensors="pt").to(cfg.DEVICE)
            out = model(**enc, labels=enc["input_ids"])
            total_log_prob += out.loss.item() * enc["input_ids"].size(-1)
            total_tokens   += enc["input_ids"].size(-1)
    return math.exp(total_log_prob / total_tokens)

def compute_emotion_accuracy(preds: list, true_labels: list) -> dict:
    """Emotion classification accuracy + per-class report."""
    acc = accuracy_score(true_labels, preds)
    report = classification_report(true_labels, preds,
                                      target_names=cfg.EMOTION_LABELS,
                                      zero_division=0, output_dict=True)
    return {"accuracy": acc, "report": report}

def compute_empathy_score(responses: list, strategies: list) -> float:
    """Keyword-based empathy score: fraction of responses containing strategy keywords."""
    score = 0
    for resp, strat in zip(responses, strategies):
        keywords = EMPATHY_KEYWORDS.get(strat, [])
        if any(kw in resp.lower() for kw in keywords):
            score += 1
    return score / len(responses) if responses else 0.0

def full_evaluation(encoder, gen_model, test_loader, gpt_tokenizer):
    """Run all metrics on test set."""
    preds_emotion, true_emotion = [], []
    generated, references, strategies = [], [], []

    encoder.eval(); gen_model.eval()
    with torch.no_grad():
        for i, batch in enumerate(test_loader):
            if i > 50: break  # quick eval subset
            ids  = batch["input_ids"].to(cfg.DEVICE)
            mask = batch["attention_mask"].to(cfg.DEVICE)
            tt   = batch["token_type_ids"].to(cfg.DEVICE)

            out = encoder(ids, mask, tt)
            pred_em = out["emotion_logits"].argmax(-1).cpu().tolist()
            preds_emotion.extend(pred_em)
            true_emotion.extend(batch["emotion_label"].tolist())

    bleu_score    = compute_bleu(generated[:50], references[:50]) if generated else 0
    em_metrics    = compute_emotion_accuracy(preds_emotion, true_emotion)
    empathy_score = compute_empathy_score(generated, strategies)

    print(f"\n{'='*40}")
    print(f"Emotion Accuracy: {em_metrics['accuracy']:.4f}")
    print(f"BLEU Score:       {bleu_score:.2f}")
    print(f"Empathy Score:    {empathy_score:.4f}")
    print(f"{'='*40}")
    return em_metrics, bleu_score, empathy_score
```

```python
%%writefile /content/eqnet_plus/modules/eqnet_model.py
import torch
import torch.nn as nn
from transformers import BertModel

import sys
# Add the parent directory of eqnet_plus to sys.path
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

from eqnet_plus.config import cfg
from eqnet_plus.modules.intensity_model import IntensityHead
from eqnet_plus.modules.empathy_model import EmpathyHead

class EQNetPlus(nn.Module):
    """
    EQ-Net+: Shared BERT encoder with 3 task heads.
    Emotion Detection | Intensity Estimation | Empathy Strategy
    """
    def __init__(self):
        super().__init__()
        # Shared backbone
        self.bert = BertModel.from_pretrained(cfg.ENCODER_MODEL)
        self.dropout = nn.Dropout(0.3)
        hidden = self.bert.config.hidden_size  # 768

        # Task heads
        self.emotion_head = nn.Sequential(
            nn.Linear(hidden, 256), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(256, cfg.NUM_EMOTIONS)
        )
        self.intensity_head = IntensityHead(hidden)
        self.empathy_head   = EmpathyHead(hidden)

    def forward(self, input_ids, attention_mask, token_type_ids=None,
               intensity_ids=None):
        # Shared encoding
        bert_out = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        cls_emb = self.dropout(bert_out.last_hidden_state[:, 0, :])

        # Individual heads
        emotion_logits   = self.emotion_head(cls_emb)
        intensity_logits = self.intensity_head(cls_emb)

        # Empathy head uses predicted intensity if not provided
        if intensity_ids is None:
            intensity_ids = intensity_logits.argmax(dim=-1)
        empathy_logits = self.empathy_head(cls_emb, intensity_ids)

        return {
            "emotion_logits":   emotion_logits,
            "intensity_logits": intensity_logits,
            "empathy_logits":   empathy_logits,
            "cls_emb":          cls_emb
        }


def compute_multitask_loss(outputs, labels):
    """Weighted multi-task loss."""
    ce = nn.CrossEntropyLoss()
    l_emotion   = ce(outputs["emotion_logits"],   labels["emotion"])
    l_intensity = ce(outputs["intensity_logits"], labels["intensity"])
    l_empathy   = ce(outputs["empathy_logits"],   labels["strategy"])

    total = (cfg.LOSS_W_EMOTION   * l_emotion +
             cfg.LOSS_W_INTENSITY * l_intensity +
             cfg.LOSS_W_STRATEGY  * l_empathy)
    return total, l_emotion, l_intensity, l_empathy

if __name__ == "__main__":
    # Ensure sys.path is correctly configured for local imports
    import sys
    if '/content' not in sys.path:
        sys.path.insert(0, '/content')

    # Force reload of eqnet_plus.config to pick up any changes
    import importlib
    if 'eqnet_plus.config' in sys.modules:
        importlib.reload(sys.modules['eqnet_plus.config'])
    from eqnet_plus.config import cfg # Re-import cfg if running directly

    model = EQNetPlus().to(cfg.DEVICE)
    print(f"EQNet+ params: {sum(p.numel() for p in model.parameters()):,}")
    # Quick forward pass test
    dummy_ids  = torch.randint(0, 30522, (2, cfg.MAX_LEN)).to(cfg.DEVICE)
    dummy_mask = torch.ones(2, cfg.MAX_LEN, dtype=torch.long).to(cfg.DEVICE)
    out = model(dummy_ids, dummy_mask)
    for k, v in out.items():
        print(f"{k}: {v.shape}")
```

```python
%%writefile /content/eqnet_plus/modules/dataset.py
import torch
from torch.utils.data import Dataset, DataLoader
import pandas as pd
import json
from transformers import BertTokenizer
import sys
import os
from pathlib import Path

# Add the parent directory of eqnet_plus to sys.path
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

# Import cfg from the project's config
from eqnet_plus.config import cfg

# Load label maps
label_maps_path = cfg.DATA_PROC / "label_maps.json"
# Added robust check for label_maps.json existence
if not label_maps_path.exists():
    print(f"Error: label_maps.json not found at {label_maps_path}. Please ensure label_encoder.py has been run successfully.")
    # Fallback with default labels to allow module creation, but actual execution without maps will fail
    MAPS = {
        "emotion2id": {e: i for i, e in enumerate(cfg.EMOTION_LABELS)},
        "id2emotion": {i: e for i, e in enumerate(cfg.EMOTION_LABELS)},
        "intensity2id": {l: i for i, l in enumerate(cfg.INTENSITY_LABELS)},
        "strategy2id": {s: i for i, s in enumerate(cfg.EMPATHY_STRATEGIES)},
    }
else:
    with open(label_maps_path, "r") as f:
        MAPS = json.load(f)

# Initialize tokenizer
tokenizer = BertTokenizer.from_pretrained(cfg.ENCODER_MODEL)

class EQNetDataset(Dataset):
    def __init__(self, csv_path):
        # Using keep_default_na=False to prevent empty strings from being interpreted as NaN
        # and then being dropped by df.dropna() or causing issues with string operations.
        # fillna("") after loading ensures all string columns have actual strings.
        self.df = pd.read_csv(csv_path, keep_default_na=False).fillna("")

        # Optional: Add a check if the dataframe is empty after loading and filling NAs
        if self.df.empty:
            print(f"Warning: DataFrame for {csv_path} is empty after loading and filling NAs.")

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        # Ensure 'utterance_clean' and 'utterance' columns exist, or handle gracefully
        utterance = str(row.get("utterance_clean", ""))
        response  = str(row.get("utterance", ""))

        enc = tokenizer(
            utterance,
            max_length=cfg.MAX_LEN,
            padding="max_length",
            truncation=True,
            return_tensors="pt"
        )

        dec = tokenizer(
            response,
            max_length=cfg.MAX_LEN,
            padding="max_length",
            truncation=True,
            return_tensors="pt"
        )

        # Labels - ensure they are in lowercase and handle potential missing keys
        # Use .get() with a default value to prevent KeyError if a label is not found
        emotion = MAPS["emotion2id"].get(str(row.get("emotion", "")).lower(), 0)
        intensity = MAPS["intensity2id"].get(str(row.get("intensity", "")).lower(), 0)
        strategy = MAPS["strategy2id"].get(str(row.get("strategy", "")).lower(), 0)

        sample = {
            "input_ids": enc["input_ids"].squeeze(),
            "attention_mask": enc["attention_mask"].squeeze(),
            # Always include token_type_ids as train.py expects it for Bert models
            "token_type_ids": enc["token_type_ids"].squeeze() if "token_type_ids" in enc else torch.zeros_like(enc["input_ids"].squeeze()),
            "decoder_ids": dec["input_ids"].squeeze(),
            "emotion_label": torch.tensor(emotion, dtype=torch.long),
            "intensity_label": torch.tensor(intensity, dtype=torch.long),
            "strategy_label": torch.tensor(strategy, dtype=torch.long)
        }
        return sample

def get_dataloaders():
    # Ensure processed data files exist before attempting to load
    train_proc_path = cfg.DATA_PROC / "train_proc.csv"
    val_proc_path   = cfg.DATA_PROC / "val_proc.csv"
    test_proc_path  = cfg.DATA_PROC / "test_proc.csv"

    if not train_proc_path.exists():
        raise FileNotFoundError(f"Processed train data not found at: {train_proc_path}. Please ensure preprocessor.py has been executed.")
    if not val_proc_path.exists():
        raise FileNotFoundError(f"Processed validation data not found at: {val_proc_path}. Please ensure preprocessor.py has been executed.")
    if not test_proc_path.exists():
        raise FileNotFoundError(f"Processed test data not found at: {test_proc_path}. Please ensure preprocessor.py has been executed.")

    train_ds = EQNetDataset(str(train_proc_path))
    val_ds   = EQNetDataset(str(val_proc_path))
    test_ds  = EQNetDataset(str(test_proc_path))

    train_loader = DataLoader(
        train_ds,
        batch_size=cfg.BATCH_SIZE,
        shuffle=True,
        num_workers=0 # Set to 0 for Colab to avoid issues with multiprocessing
    )

    val_loader = DataLoader(
        val_ds,
        batch_size=cfg.BATCH_SIZE,
        shuffle=False,
        num_workers=0
    )

    test_loader = DataLoader(
        test_ds,
        batch_size=cfg.BATCH_SIZE,
        shuffle=False,
        num_workers=0
    )

    print(f"Train batches: {len(train_loader)} | Val: {len(val_loader)} | Test: {len(test_loader)}")
    return train_loader, val_loader, test_loader

if __name__ == "__main__":
    print("Running dataset.py for testing purposes...")
    # This block assumes the necessary raw/processed data and label maps have been created by previous steps (data_loader.py, preprocessor.py, label_encoder.py)
    # If not, this standalone execution might fail, which is expected for testing modules in isolation.
    try:
        train_loader, val_loader, test_loader = get_dataloaders()
        batch = next(iter(train_loader))
        print("\nSample batch shapes:")
        for k, v in batch.items():
            print(f"{k}: {v.shape}")
    except Exception as e:
        print(f"Error during standalone dataset.py test: {e}")
        print("Please ensure data_loader.py, label_encoder.py, and preprocessor.py have been executed to create necessary files.")
```

```python
# modules/personalization.py
import json
from pathlib import Path
from collections import Counter, defaultdict
from datetime import datetime
import os # Import os module to create directories

# Corrected PROFILE_PATH to explicitly use the full path expected
# by the overall project structure, assuming the script is run from /content
PROFILE_PATH = Path("/content/eqnet_plus/data/processed/user_profiles.json")

def load_profiles() -> dict:
    if PROFILE_PATH.exists():
        with open(PROFILE_PATH) as f:
            return json.load(f)
    return {}

def save_profiles(profiles: dict):
    # Ensure the directory exists before saving the file
    os.makedirs(PROFILE_PATH.parent, exist_ok=True)
    with open(PROFILE_PATH, "w") as f:
        json.dump(profiles, f, indent=2)

def get_or_create_profile(user_id: str) -> dict:
    profiles = load_profiles()
    if user_id not in profiles:
        profiles[user_id] = {
            "emotion_history":    [],
            "preferred_length":   "medium",  # short/medium/long
            "dominant_emotion":   None,
            "interaction_count":  0,
            "last_seen":          str(datetime.now())
        }
        save_profiles(profiles)
    return profiles[user_id]

def update_profile(user_id: str, emotion: str, utterance_len: int):
    profiles = load_profiles()
    profile  = profiles.get(user_id, get_or_create_profile(user_id))

    profile["emotion_history"].append(emotion)
    profile["interaction_count"] += 1
    profile["last_seen"] = str(datetime.now())

    # Update dominant emotion (most frequent in last 10 turns)
    recent = profile["emotion_history"][-10:]
    profile["dominant_emotion"] = Counter(recent).most_common(1)[0][0]

    # Infer preferred response length from user message length
    if utterance_len < 20:
        profile["preferred_length"] = "short"
    elif utterance_len > 60:
        profile["preferred_length"] = "long"

    profiles[user_id] = profile
    save_profiles(profiles)
    return profile

def build_personalized_prompt(utterance: str, user_id: str,
                              emotion: str, strategy: str) -> str:
    """Build enriched prompt for the generator."""
    profile = get_or_create_profile(user_id)
    dominant = profile.get("dominant_emotion", emotion) or emotion
    length   = profile.get("preferred_length", "medium")

    length_hint = {
        "short":  "Reply briefly in 1-2 sentences.",
        "medium": "Reply in 2-3 sentences.",
        "long":   "Reply thoughtfully in 3-5 sentences.",
    }[length]

    prompt = (
        f"[Emotion: {emotion}] [Dominant: {dominant}] [Strategy: {strategy}]\n"
        f"User: {utterance}\n"
        f"Empathetic Assistant ({length_hint}): "
    )
    return prompt

if __name__ == "__main__":
    # Ensure the /content/eqnet_plus/data/processed directory exists for the test
    os.makedirs(PROFILE_PATH.parent, exist_ok=True)

    # Test
    uid = "user_001"
    update_profile(uid, "sad", 25)
    update_profile(uid, "anxious", 18)
    prompt = build_personalized_prompt(
        "I'm feeling overwhelmed at work", uid, "anxious", "reassurance")
    print(prompt)
```

```python
%%writefile eqnet_plus/api/main.py
import sys
import os

# Ensure /content is in sys.path to discover the eqnet_plus package
if '/content' not in sys.path:
    sys.path.insert(0, '/content')
    print("Added /content to sys.path.")

# Invalidate caches for eqnet_plus and its submodules to force a fresh import
import importlib
for module_name in list(sys.modules.keys()):
    if module_name.startswith('eqnet_plus'):
        del sys.modules[module_name]
importlib.invalidate_caches()

from fastapi import FastAPI, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
import torch, json
from transformers import BertTokenizer

# Corrected import paths to use the 'eqnet_plus' package prefix
from eqnet_plus.modules.eqnet_model import EQNetPlus
from eqnet_plus.modules.response_model import EmpathyResponseGenerator
from eqnet_plus.modules.cause_model import identify_cause
from eqnet_plus.modules.preprocessor import clean_text
from eqnet_plus.modules.personalization import (
    update_profile, build_personalized_prompt, get_or_create_profile)
from eqnet_plus.config import cfg

app = FastAPI(title="EQ-Net+ API", version="1.0.0")
app.add_middleware(CORSMiddleware, allow_origins=["*"],
                    allow_methods=["*"], allow_headers=["* "])

# -- Load models at startup
# Adjusted path to be absolute or relative to the current working directory if not running from root
with open("/content/eqnet_plus/data/processed/label_maps.json") as f:
    MAPS = json.load(f)

tokenizer = BertTokenizer.from_pretrained(cfg.ENCODER_MODEL)
encoder   = EQNetPlus().to(cfg.DEVICE)
generator = EmpathyResponseGenerator().to(cfg.DEVICE)

encoder.load_state_dict(
    torch.load("/content/eqnet_plus/models/eqnet_encoder_best.pt", map_location=cfg.DEVICE))
generator.load_state_dict(
    torch.load("/content/eqnet_plus/models/eqnet_generator_best.pt", map_location=cfg.DEVICE))
encoder.eval(); generator.eval()
print("✓ Models loaded")

# -- Pydantic schemas
class AnalyzeRequest(BaseModel):
    utterance: str
    user_id:   str = "anonymous"

class AnalyzeResponse(BaseModel):
    emotion:   str
    intensity: str
    strategy:  str
    cause:     str
    confidence: float

class RespondRequest(BaseModel):
    utterance: str
    user_id:   str = "anonymous"

class RespondResponse(BaseModel):
    response:  str
    emotion:   str
    intensity: str
    strategy:  str
    cause:     str

# -- Helper
def encode_utterance(text: str):
    text = clean_text(text)
    enc  = tokenizer(text, max_length=cfg.MAX_LEN, padding="max_length",
                     truncation=True, return_tensors="pt")
    return {k: v.to(cfg.DEVICE) for k, v in enc.items()}

def infer_all(text: str):
    enc = encode_utterance(text)
    with torch.no_grad():
        out = encoder(enc["input_ids"], enc["attention_mask"],
                       enc["token_type_ids"])
    em_id  = out["emotion_logits"].argmax(-1).item()
    in_id  = out["intensity_logits"].argmax(-1).item()
    st_id  = out["empathy_logits"].argmax(-1).item()
    conf   = torch.softmax(out["emotion_logits"], dim=-1).max().item()
    emotion  = MAPS["id2emotion"][str(em_id)]
    intensity= cfg.INTENSITY_LABELS[in_id]
    strategy = cfg.EMPATHY_STRATEGIES[st_id]
    cause    = identify_cause(text)["cause"]
    return emotion, intensity, strategy, cause, conf, em_id, st_id

# -- Routes
@app.post("/analyze", response_model=AnalyzeResponse)
async def analyze(req: AnalyzeRequest):
    emotion, intensity, strategy, cause, conf, _, _ = infer_all(req.utterance)
    return AnalyzeResponse(emotion=emotion, intensity=intensity,
                            strategy=strategy, cause=cause, confidence=conf)

@app.post("/respond", response_model=RespondResponse)
async def respond(req: RespondRequest):
    emotion, intensity, strategy, cause, conf, em_id, st_id = infer_all(req.utterance)
    update_profile(req.user_id, emotion, len(req.utterance.split()))
    response = generator.generate_response(req.utterance, em_id, st_id)
    return RespondResponse(response=response, emotion=emotion,
                            intensity=intensity, strategy=cause)

@app.get("/health")
def health():
    return {"status": "ok", "device": cfg.DEVICE}
```

```python
%%writefile /content/eqnet_plus/modules/cause_model.py
import spacy

nlp = spacy.load("en_core_web_sm")

def extract_cause_spacy(text: str) -> str:
    lower_text = text.lower()

    # Rule 1: The cause comes AFTER these words
    if "because" in lower_text:
        return text.split("because", 1)[1].strip(" .")
    if "due to" in lower_text:
        return text.split("due to", 1)[1].strip(" .")
    if "since" in lower_text:
        return text.split("since", 1)[1].strip(" .")
    if "when" in lower_text:
        return text.split("when", 1)[1].strip(" .")

    # Rule 2: The cause comes BEFORE these words
    if "therefore" in lower_text:
        return text.split("therefore", 1)[0].strip(" .")
    if "and" in lower_text:
        return text.split("and", 1)[0].strip(" .")

    return text


def identify_cause(text: str) -> dict:
    doc = nlp(text)
    entities = [(ent.text, ent.label_) for ent in doc.ents]
    cause_phrase = extract_cause_spacy(text)
    return {"cause": cause_phrase, "entities": entities}


if __name__ == "__main__":
    test_texts = [
        "I am so sad because I lost my job today.",
        "I feel anxious when I have to speak in public.",
        "She broke up with me and I'm devastated.",
        "The picnic was canceled due to the heavy rain.",
        "He missed the flight since his alarm didn't go off.",
        "She won the award therefore she felt extremely proud.",
    ]

    for t in test_texts:
        result = identify_cause(t)
        print(f"Text: {t}")
        print(f"Cause: {result['cause']} | Entities: {result['entities']}\n")
```

```python
%%writefile evaluation/metrics.py
import torch
import math
from transformers import GPT2LMHeadModel, GPT2Tokenizer

import sys
import os
from pathlib import Path # Import Path for clearer path handling
import importlib

# Add the parent directory of eqnet_plus to sys.path
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

print(f"DEBUG (metrics.py): Current working directory: {os.getcwd()}")
print(f"DEBUG (metrics.py): sys.path is: {sys.path}")
print(f"DEBUG (metrics.py): Does /content/eqnet_plus exist? {os.path.exists('/content/eqnet_plus')}")
print(f"DEBUG (metrics.py): Does /content/eqnet_plus/config.py exist? {os.path.exists('/content/eqnet_plus/config.py')}")

# --- Robust import for config --- #
try:
    # Force reload config.py to ensure latest settings are used
    if 'eqnet_plus.config' in sys.modules:
        importlib.reload(sys.modules['eqnet_plus.config'])
    from eqnet_plus.config import cfg
    print(f"DEBUG (metrics.py): Config imported successfully from eqnet_plus.config. cfg.DEVICE: {cfg.DEVICE}")
except ImportError as e:
    print(f"WARNING (metrics.py): Could not import eqnet_plus.config ({e}). Defining a mock config for evaluation metrics.")
    class MockConfig:
        # Minimal config needed for metrics.py, adjust as necessary
        DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
        EMOTION_LABELS = [
            "admiring", "afraid", "angry", "annoyed", "anticipating",
            "anxious", "apprehensive", "ashamed", "caring", "confident",
            "content", "devastated", "disappointed", "disgusted", "embarrassed",
            "excited", "faithful", "furious", "grateful", "guilty",
            "hopeful", "impressed", "jealous", "joyful", "lonely",
            "nostalgic", "prepared", "proud", "sad", "sentimental",
            "surprised", "terrified"
        ]
        NUM_EMOTIONS = 32 # Added NUM_EMOTIONS to mock config
        DECODER_MODEL = "microsoft/DialoGPT-small" # Needed for tokenizer
    cfg = MockConfig()
# --- End robust import for config --- #


# --- Robust import for 'evaluate' and sacrebleu --- #
try:
    from evaluate import load
    bleu = load("sacrebleu")
    sacrebleu_present = True
except ImportError:
    print("WARNING (metrics.py): The 'evaluate' library (or 'sacrebleu' component) is not installed.")
    print("Please install it using: !pip install evaluate sacrebleu")
    class MockBLEU: # Define a mock object to prevent further errors
        def compute(self, predictions, references):
            print("BLEU computation skipped due to missing 'evaluate' library.")
            return {"score": 0.0}
    bleu = MockBLEU()
    sacrebleu_present = False
# --- End robust import for 'evaluate' and sacrebleu --- #


# --- Robust import for 'sklearn.metrics' --- #
try:
    from sklearn.metrics import accuracy_score, classification_report
    sklearn_present = True
except ImportError:
    print("WARNING (metrics.py): The 'scikit-learn' library is not installed.")
    print("Please install it using: !pip install scikit-learn")
    # Define mock functions to prevent further errors
    def accuracy_score(y_true, y_pred): # pylint: disable=unused-argument
        print("Accuracy computation skipped due to missing 'scikit-learn' library.")
        return 0.0
    def classification_report(y_true, y_pred, target_names, zero_division, output_dict): # pylint: disable=unused-argument
        print("Classification report skipped due to missing 'scikit-learn' library.")
        if output_dict:
            return {"accuracy": 0.0, "report": "scikit-learn not available"}
        return "scikit-learn not available"
    sklearn_present = False
# --- End robust import for 'sklearn.metrics' --- #


# Empathy keywords per strategy
EMPATHY_KEYWORDS = {
    "validation":     ["understand", "makes sense", "valid", "feel"],
    "encouragement":  ["believe in you", "keep going", "strong", "you can"],
    "reassurance":    ["alright", "okay", "be fine", "safe"],
    "acknowledgement":["that", "wonderful", "great", "happy for you"],
    "suggestion":     ["try", "maybe", "could", "consider"],
}

def compute_bleu(predictions: list, references: list) -> float:
    """Compute corpus BLEU score."""
    if not sacrebleu_present or not predictions or not references:
        return 0.0
    # References should be List[List[str]] for sacrebleu.compute
    references_formatted = [[r] for r in references] if not all(isinstance(elem, list) for elem in references) else references
    result = bleu.compute(predictions=predictions, references=references_formatted)
    return result["score"]

def compute_perplexity(model_or_lmheadmodel, tokenizer_gpt, texts: list) -> float:
    """Compute average perplexity on a list of texts."""
    if not texts:
        return float('inf')
    # If the model is an EmpathyResponseGenerator instance, use its internal GPT model
    if hasattr(model_or_lmheadmodel, 'gpt') and isinstance(model_or_lmheadmodel.gpt, GPT2LMHeadModel):
        model_for_perplexity = model_or_lmheadmodel.gpt
    elif isinstance(model_or_lmheadmodel, GPT2LMHeadModel):
        model_for_perplexity = model_or_lmheadmodel
    else:
        print("WARNING (metrics.py): Perplexity model not a GPT2LMHeadModel or EmpathyResponseGenerator. Returning inf.")
        return float('inf')

    model_for_perplexity.eval()

    total_loss = 0
    total_tokens = 0

    with torch.no_grad():
        for text in texts:
            if not text:
                continue # Skip empty texts
            try:
                enc = tokenizer_gpt(text, return_tensors="pt").to(cfg.DEVICE)
                if enc["input_ids"].numel() == 0: # Check if tokenizer returned empty input_ids
                    continue
                outputs = model_for_perplexity(**enc, labels=enc["input_ids"])
                loss = outputs.loss.item()
                total_loss += loss * enc["input_ids"].shape[1]
                total_tokens += enc["input_ids"].shape[1]
            except Exception as e:
                print(f"WARNING (metrics.py): Error processing text for perplexity: {text[:50]}... Error: {e}")
                continue

    if total_tokens == 0:
        return float('inf') # Return inf for empty or unprocessable texts

    avg_loss = total_loss / total_tokens
    return math.exp(avg_loss)

def compute_emotion_accuracy(preds: list, true_labels: list) -> dict:
    """Emotion classification accuracy + per-class report."""
    if not sklearn_present or not preds or not true_labels:
        return {"accuracy": 0.0, "report": "scikit-learn not available or no data"}

    acc = accuracy_score(true_labels, preds)
    # FIX: Add 'labels' parameter to classification_report for robustness
    # Ensure target_names matches the number of unique labels found, or total possible labels
    unique_true_labels = sorted(list(set(true_labels)))
    if len(unique_true_labels) > 0 and max(unique_true_labels) < cfg.NUM_EMOTIONS: # Simple check for valid labels
        labels_to_use = list(range(cfg.NUM_EMOTIONS))
        target_names_to_use = cfg.EMOTION_LABELS
    else:
        # Fallback if labels don't align with full config labels (e.g., in partial data scenarios)
        labels_to_use = unique_true_labels
        target_names_to_use = [str(x) for x in unique_true_labels] # Use string representation if names unavailable

    try:
        report = classification_report(true_labels, preds,
                                          labels=labels_to_use, # Explicitly pass labels
                                          target_names=target_names_to_use,
                                          zero_division=0, output_dict=True)
    except ValueError as ve:
        print(f"WARNING (metrics.py): Error in classification_report: {ve}. Falling back to accuracy only.")
        report = {"accuracy": acc, "report": f"Error generating full report: {ve}"}

    return {"accuracy": acc, "report": report}

def compute_empathy_score(responses: list, strategies: list) -> float:
    """Keyword-based empathy score: fraction of responses containing strategy keywords."""
    if not responses or not strategies:
        return 0.0
    score = 0
    for resp, strat in zip(responses, strategies):
        keywords = EMPATHY_KEYWORDS.get(strat, [])
        if any(kw in resp.lower() for kw in keywords):
            score += 1
    return score / len(responses)

def full_evaluation(encoder, gen_model, test_loader, gpt_tokenizer_obj):
    """Run all metrics on test set."""
    preds_emotion, true_emotion = [], []
    generated, references, strategies = [], [], []

    encoder.eval(); gen_model.eval()
    with torch.no_grad():
        for i, batch in enumerate(test_loader):
            if i > 50: break  # quick eval subset for speed

            ids  = batch["input_ids"].to(cfg.DEVICE)
            mask = batch["attention_mask"].to(cfg.DEVICE)
            tt   = batch["token_type_ids"].to(cfg.DEVICE)

            # Encoder predictions
            out = encoder(ids, mask, tt)
            pred_em_ids = out["emotion_logits"].argmax(-1).cpu().tolist()
            preds_emotion.extend(pred_em_ids)
            true_emotion.extend(batch["emotion_label"].tolist())

            # Generator responses
            # Note: We need actual emotion and strategy IDs for generation, not just logits
            # For evaluation, we'll use the true labels for the generation context
            true_em_ids = batch["emotion_label"].tolist()
            true_st_ids = batch["strategy_label"].tolist()
            original_utterances = [gpt_tokenizer_obj.decode(ids[j], skip_special_tokens=True) for j in range(ids.size(0))]

            for j in range(ids.size(0)):
                # Generate response for each item in batch
                generated_text = gen_model.generate_response(
                    original_utterances[j], true_em_ids[j], true_st_ids[j]
                )
                generated.append(generated_text)
                # Ensure references are list of lists for sacrebleu
                references.append([gpt_tokenizer_obj.decode(batch["decoder_ids"][j], skip_special_tokens=True)])
                strategies.append(cfg.EMPATHY_STRATEGIES[true_st_ids[j]])


    bleu_score    = compute_bleu(generated, references)
    em_metrics    = compute_emotion_accuracy(preds_emotion, true_emotion)
    empathy_score = compute_empathy_score(generated, strategies)
    perplexity    = compute_perplexity(gen_model, gpt_tokenizer_obj, generated)

    print(f"\n{'='*40}")
    print(f"Emotion Accuracy: {em_metrics['accuracy']:.4f}")
    print(f"BLEU Score:       {bleu_score:.2f}")
    print(f"Empathy Score:    {empathy_score:.4f}")
    print(f"Perplexity:       {perplexity:.2f}")
    print(f"{'='*40}")
    return em_metrics, bleu_score, empathy_score, perplexity
```

```python
import sys
import os
import math
import torch
from evaluate import load
from sklearn.metrics import accuracy_score, classification_report

# Add the parent directory of eqnet_plus to sys.path
if '/content' not in sys.path:
    sys.path.insert(0, '/content')

from eqnet_plus.config import cfg

# Load BLEU
bleu = load("sacrebleu")


# ===========================================================
# Empathy Keywords
# ===========================================================

EMPATHY_KEYWORDS = {
    "validation": [
        "understand",
        "makes sense",
        "valid",
        "feel"
    ],

    "encouragement": [
        "keep going",
        "you can",
        "believe",
        "strong"
    ],

    "reassurance": [
        "okay",
        "alright",
        "safe",
        "be fine"
    ],

    "acknowledgement": [
        "happy",
        "great",
        "wonderful",
        "glad"
    ],

    "suggestion": [
        "try",
        "maybe",
        "consider",
        "could"
    ]
}


# ===========================================================
# BLEU
# ===========================================================

def compute_bleu(predictions, references):

    if len(predictions) == 0:
        return 0.0

    references = [[r] for r in references]

    score = bleu.compute(
        predictions=predictions,
        references=references
    )

    return score["score"]


# ===========================================================
# Perplexity
# ===========================================================

def compute_perplexity(model, tokenizer, texts):

    model.eval()

    total_loss = 0
    total_tokens = 0

    with torch.no_grad():

        for text in texts:

            inputs = tokenizer(
                text,
                return_tensors="pt",
                truncation=True,
                max_length=cfg.MAX_LEN
            )

            inputs = {k: v.to(cfg.DEVICE) for k, v in inputs.items()}

            outputs = model(
                **inputs,
                labels=inputs["input_ids"]
            )

            loss = outputs.loss.item()

            total_loss += loss * inputs["input_ids"].shape[1]
            total_tokens += inputs["input_ids"].shape[1]

    if total_tokens == 0:
        return 0

    avg_loss = total_loss / total_tokens

    return math.exp(avg_loss)


# ===========================================================
# Emotion Accuracy
# ===========================================================

def compute_emotion_accuracy(preds, labels):

    accuracy = accuracy_score(labels, preds)

    report = classification_report(
        labels,
        preds,
        target_names=cfg.EMOTION_LABELS,
        output_dict=True,
        zero_division=0
    )

    return {
        "accuracy": accuracy,
        "report": report
    }


# ===========================================================
# Empathy Score
# ===========================================================

def compute_empathy_score(responses, strategies):

    if len(responses) == 0:
        return 0

    score = 0

    for response, strategy in zip(responses, strategies):

        keywords = EMPATHY_KEYWORDS.get(strategy, [])

        response = response.lower()

        if any(keyword in response for keyword in keywords):
            score += 1

    return score / len(responses)


# ===========================================================
# Response Relevance
# ===========================================================

def compute_response_relevance(predictions, references):

    if len(predictions) == 0:
        return 0

    score = 0

    for pred, ref in zip(predictions, references):

        pred_words = set(pred.lower().split())
        ref_words = set(ref.lower().split())

        if len(ref_words) == 0:
            continue

        overlap = pred_words.intersection(ref_words)

        score += len(overlap) / len(ref_words)

    return score / len(predictions)


# ===========================================================
# Full Evaluation
# ===========================================================

def full_evaluation(
    encoder,
    generator,
    tokenizer,
    test_loader
):

    encoder.eval()
    generator.eval()

    emotion_predictions = []
    emotion_labels = []

    generated_responses = []
    reference_responses = []
    strategies = []

    with torch.no_grad():

        for batch in test_loader:

            input_ids = batch["input_ids"].to(cfg.DEVICE)

            attention_mask = batch["attention_mask"].to(cfg.DEVICE)

            token_type_ids = batch["token_type_ids"].to(cfg.DEVICE)

            outputs = encoder(
                input_ids,
                attention_mask,
                token_type_ids
            )

            emotion_logits = outputs["emotion_logits"]

            preds = torch.argmax(
                emotion_logits,
                dim=1
            )

            emotion_predictions.extend(
                preds.cpu().numpy().tolist()
            )

            emotion_labels.extend(
                batch["emotion_label"].numpy().tolist()
            )

            # -------------------------
            # Generate Responses
            # -------------------------

            generated_ids = generator.generate(

                input_ids=input_ids,

                attention_mask=attention_mask,

                max_length=60,

                do_sample=True,

                temperature=0.7,

                top_p=0.9,

                pad_token_id=tokenizer.eos_token_id

            )

            preds_text = tokenizer.batch_decode(

                generated_ids,

                skip_special_tokens=True

            )

            refs_text = tokenizer.batch_decode(

                input_ids,

                skip_special_tokens=True

            )

            generated_responses.extend(preds_text)

            reference_responses.extend(refs_text)

            if "strategy" in batch:

                strategies.extend(batch["strategy"])

            else:

                strategies.extend(

                    ["validation"] * len(preds_text)

                )

    emotion_metrics = compute_emotion_accuracy(

        emotion_predictions,

        emotion_labels

    )

    bleu_score = compute_bleu(

        generated_responses,

        reference_responses

    )

    empathy_score = compute_empathy_score(

        generated_responses,

        strategies

    )

    relevance_score = compute_response_relevance(

        generated_responses,

        reference_responses

    )

    perplexity = compute_perplexity(

        generator,

        tokenizer,

        generated_responses
    )

    print("\n")
    print("=" * 60)

    print(f"Emotion Accuracy  : {emotion_metrics['accuracy']:.4f}")

    print(f"BLEU Score        : {bleu_score:.2f}")

    print(f"Perplexity        : {perplexity:.2f}")

    print(f"Empathy Score     : {empathy_score:.4f}")

    print(f"Response Relevance: {relevance_score:.4f}")

    print("=" * 60)

    return {

        "Emotion Accuracy": emotion_metrics["accuracy"],

        "BLEU": bleu_score,

        "Perplexity": perplexity,

        "Empathy Score": empathy_score,

        "Response Relevance": relevance_score,

        "Classification Report": emotion_metrics["report"]

    }
```

```python
%%writefile /content/eqnet_plus/modules/personalization.py
# modules/personalization.py
import json
from pathlib import Path
from collections import Counter, defaultdict
from datetime import datetime
import os # Import os module to create directories

PROFILE_PATH = Path("/content/eqnet_plus/data/processed/user_profiles.json")

def load_profiles() -> dict:
    if PROFILE_PATH.exists():
        with open(PROFILE_PATH) as f:
            return json.load(f)
    return {}

def save_profiles(profiles: dict):
    # Ensure the directory exists before saving the file
    os.makedirs(PROFILE_PATH.parent, exist_ok=True)
    with open(PROFILE_PATH, "w") as f:
        json.dump(profiles, f, indent=2)

def get_or_create_profile(user_id: str) -> dict:
    profiles = load_profiles()
    if user_id not in profiles:
        profiles[user_id] = {
            "emotion_history":    [],
            "preferred_length":   "medium",  # short/medium/long
            "dominant_emotion":   None,
            "interaction_count":  0,
            "last_seen":          str(datetime.now())
        }
        save_profiles(profiles)
    return profiles[user_id]

def update_profile(user_id: str, emotion: str, utterance_len: int):
    profiles = load_profiles()
    profile  = profiles.get(user_id, get_or_create_profile(user_id))

    profile["emotion_history"].append(emotion)
    profile["interaction_count"] += 1
    profile["last_seen"] = str(datetime.now())

    # Update dominant emotion (most frequent in last 10 turns)
    recent = profile["emotion_history"][-10:]
    profile["dominant_emotion"] = Counter(recent).most_common(1)[0][0]

    # Infer preferred response length from user message length
    if utterance_len < 20:
        profile["preferred_length"] = "short"
    elif utterance_len > 60:
        profile["preferred_length"] = "long"

    profiles[user_id] = profile
    save_profiles(profiles)
    return profile

def build_personalized_prompt(utterance: str, user_id: str,
                              emotion: str, strategy: str) -> str:
    """Build enriched prompt for the generator."""
    profile = get_or_create_profile(user_id)
    dominant = profile.get("dominant_emotion", emotion) or emotion
    length   = profile.get("preferred_length", "medium")

    length_hint = {
        "short":  "Reply briefly in 1-2 sentences.",
        "medium": "Reply in 2-3 sentences.",
        "long":   "Reply thoughtfully in 3-5 sentences.",
    }[length]

    prompt = (
        f"[Emotion: {emotion}] [Dominant: {dominant}] [Strategy: {strategy}]\n"
        f"User: {utterance}\n"
        f"Empathetic Assistant ({length_hint}): "
    )
    return prompt
```
